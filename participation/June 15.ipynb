```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
tips = sns.load_dataset("tips")
print(tips)
sns.set_style("whitegrid")
sns.lineplot(data=tips, x="total_bill", y="tip")
tipgraph = sns.lineplot(data=tips, x='total_bill', y='tip')
tipgraph.set(title='Tip vs Total Bill', xlabel='Total Bill($)', ylabel='Tip amount($)')
```

```python
tips = sns.load_dataset("tips")
print(tips)
sns.set_style("dark")
tipgraph = sns.lineplot(data=tips, x='total_bill', y='tip',
                        color='hotpink',
                        linestyle='--',
                        linewidth=3,
                        marker='o',
                        markerfacecolor='indigo')
tipgraph.set(title='Tip vs Total Bill', xlabel='Total Bill($)', ylabel='Tip Amount($)')
```

```python
tipgraph = sns.scatterplot(data=tips, x='total_bill',
                           y='tip', style='time', hue='day',
                           palette=['purple',
                                    'hotpink', 'deepskyblue',
                                    'yellowgreen'])

tipgraph.set(title='Tip vs Total Bill',
             xlabel='Total Bill ($)',
             ylabel='Tip Amount ($)')
```

```python
sns.pairplot(data=tips)
```

```python
daysplot = sns.relplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    col="day",
    kind="scatter",
    col_wrap=2
)
```

```python
import plotly.graph_objects as go   

x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
```

```python
graph =go.Figure()
graph.add_trace(go.Bar(x=x1,y=y1))

graph.update_layout(title="Pirate Scores",
                    xaxis_title="Pirates",
                    yaxis_title="Scores")

#graph.write_html("pirategraph.html")
```

```python
graph = go.Figure()
graph.add_trace(go.Scatter(x=x1, y=y1, mode='markers',    # we want points for a scatter plot
    marker=dict(
        size=15,                         # point size
        color='hotpink',                 # point colour
        opacity=1,                       # point transparency/alpha
        line=dict(width=5, color='purple')   # point outline
    )))

graph.update_layout(
    title='Interactive Pirate Plot',
    xaxis_title='Pirates',
    yaxis_title='Scores',
    width=500,
    height=500)
```

```python
from matplotlib_venn import venn2_unweighted

A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])

diagram = venn2_unweighted(
    [A, B],
    set_labels=('Fruits', 'Pies'),
    set_colors=("blue", "red"),
    alpha=0.5
)

diagram.get_label_by_id("10").set_text("\n".join(A - B))

diagram.get_label_by_id("11").set_text("\n".join(A & B))

diagram.get_label_by_id("01").set_text("\n".join(B - A))
```
